````python
def extract_data_from_card(card):

    try:
        description = card.find_element(By.CSS_SELECTOR, ".card-title__title").text

        base = card.find_element(By.CSS_SELECTOR, ".card-base")
        date = base.find_element(
            By.CSS_SELECTOR, "p.card-content__description"
        ).get_attribute("innerHTML")
        description = base.find_element(
            By.CSS_SELECTOR, "p.card-content__title"
        ).get_attribute("innerHTML")

        a_elem = base.find_element(By.CSS_SELECTOR, "a")

        # print(a_elem.get_property("innerHTML"))

        url = a_elem.get_property("href")
        # description = a_elem.get_property("aria-label")

        return dict(description=description, date=date, url=url)

    except:
        return None
````

````python
lod = []

game_cards = driver.find_elements(By.XPATH, "//div[@data-card-type='VOD_VIDEO']")

for card in tqdm(game_cards):
    data = extract_data_from_card(card)
    if data is not None:
        lod.append(data)

lod
````
