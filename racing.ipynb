```python
gplus_df["gplus_p96"] = (
    96 * gplus_df["goals_added_above_replacement"] / gplus_df["minutes_played"]
)

gplus_df.sort_values("gplus_p96", ascending=False)
```

```python
def explode_actions(action_list):
    row_dict = {}
    for action in action_list:
        action_type = action["action_type"]
        for action_value in [
            "goals_added_raw",
            # "goals_added_above_avg",
            # "count_actions",
        ]:
            row_dict[f"{action_type} {action_value}"] = action[action_value]

    return row_dict


gplus_df = decorate_player_ids(
    decorate_team_ids(
        asa_client.get_player_goals_added(
            leagues="nwsl",
            season_name="2025",
            team_ids="eV5DR6YQKn",
            # split_by_games=True,
        )
    )
)

gplus_df = pd.concat(
    [gplus_df, pd.DataFrame(gplus_df.data.apply(explode_actions).to_list())], axis=1
)
```

```python
gplus_df.loc[:, gplus_df.columns[gplus_df.columns.str.contains("goals_added_raw")]].sum(
    axis=1
)
```

```python
gplus_df["goals_added_raw_total"] = gplus_df.loc[
    :, gplus_df.columns[gplus_df.columns.str.contains("goals_added_raw")]
].sum(axis=1)
```

```python
gplus_df["goals_added_raw_p96"] = (
    96 * gplus_df["goals_added_raw_total"] / gplus_df["minutes_played"]
)
```

```python
gplus_df.sort_values("goals_added_raw_p96", ascending=False)
```

```python
team_df = asa_client.get_team_goals_added(
    leagues="nwsl", season="2025", gamestate_trunc=["-1", "0"]
)


team_df = decorate_team_ids(team_df)
team_df["xgoal_difference_per_game"] = team_df.xgoal_difference / team_df.count_games

team_df = team_df.sort_values("xgoal_difference_per_game", ascending=False).reset_index(
    drop=True
)

plot = (
    ggplot(
        team_df,
        aes(x="team", y="xgoal_difference_per_game", fill="team"),
    )
    + geom_col()
    + scale_fill_manual(nwsl_primary_colors, guide=None)
    + scale_x_discrete(limits=team_df.team)
    + geom_image(aes(image="team_logo"))
    + labs(
        y="xGD",
        title=f"Average expected goal differential by team, NWSL 2025 season through week {team_df.count_games.min()}",
        caption="""
    xGD here measures expected goals - expected goals allowed by team per game, averaged over the season.
    xGD is adjusted for the share of home games played. 
    Data: American Soccer Analysis""",
    )
)
```
