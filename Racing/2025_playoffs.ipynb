```python
from itscalledsoccer.client import AmericanSoccerAnalysis
from plotnine import *
from colour import Color
import numpy as np
import pandas as pd

asa_client = AmericanSoccerAnalysis()

from sophso import (
    decorate_team_ids,
    nwsl_primary_colors,
    geom_image,
    decorate_player_ids,
    decorate_game_ids,
)
```

```python
teams_df = asa_client.get_teams(leagues="nwsl")
```

```python
teams_df
```

```python
games_2025 = asa_client.get_games(leagues="nwsl", seasons="2025")
```

```python
playoff_teams = ["KC", "SEA", "SD", "WAS", "LOU", "POR", "NJY", "ORL"]

games_df = decorate_game_ids(
    decorate_team_ids(asa_client.get_game_xgoals(game_ids=games_2025.game_id.to_list()))
)


games_df = games_df[
    games_df.home_team.isin(playoff_teams) & games_df.away_team.isin(playoff_teams)
]

games_df["home_points"] = 0
games_df["away_points"] = 0
games_df.loc[games_df.final_score_difference > 0, "home_points"] = 3
games_df.loc[games_df.final_score_difference < 0, "away_points"] = 3
games_df.loc[games_df.final_score_difference == 0, "home_points"] = 1
games_df.loc[games_df.final_score_difference == 0, "away_points"] = 1


games_df.sort_values("home_team")
```

```python
games_df.groupby("away_team").away_points.sum().sort_values(ascending=False)
```

```python
games_df.groupby("home_team").home_points.sum().sort_values(ascending=False)
```

```python
(
    games_df.groupby("away_team").away_points.sum()
    + games_df.groupby("home_team").home_points.sum()
).sort_values(ascending=False)
```
