```python
from itscalledsoccer.client import AmericanSoccerAnalysis
from plotnine import *
from colour import Color
import numpy as np
import pandas as pd

asa_client = AmericanSoccerAnalysis()

from sophso import (
    decorate_team_ids,
    nwsl_primary_colors,
    geom_image,
    decorate_player_ids,
    decorate_game_ids,
)
```

```python
games_df = decorate_team_ids(
    asa_client.get_games(
        leagues="nwsl",
        seasons="2025",
        # status=["FullTime", "PreMatch"],
        status="FullTime",
    )
)
```

```python
games_df.date_time_utc = pd.to_datetime(games_df.date_time_utc).dt.tz_convert(
    "US/Pacific"
)
```

```python
games_df["game_day_name"] = games_df.date_time_utc.dt.day_name()
```

```python
games_df
```

```python
games_df.pivot_table(
    index="home_team", columns=["game_day_name"], values="game_id", aggfunc="count"
).sort_values("Friday", ascending=False)
```

```python
games_df.pivot_table(
    index="away_team", columns=["game_day_name"], values="game_id", aggfunc="count"
).sort_values("Friday", ascending=False)
```

```python
(
    games_df.pivot_table(
        index="home_team", columns=["game_day_name"], values="game_id", aggfunc="count"
    )
    + games_df.pivot_table(
        index="away_team", columns=["game_day_name"], values="game_id", aggfunc="count"
    )
).sort_values("Friday", ascending=False).columns
```
