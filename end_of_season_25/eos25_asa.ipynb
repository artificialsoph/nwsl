```python
def explode_actions(action_list):
    row_dict = {}
    for action in action_list:
        action_type = action["action_type"]

        action_values = list(action.keys())
        action_values.remove("action_type")
        for action_value in action_values:
            row_dict[f"{action_type} {action_value}"] = action[action_value]

    return row_dict


def decorate_goals_added(g_df):
    g_df = pd.concat(
        [g_df, pd.DataFrame(g_df.data.apply(explode_actions).to_list())], axis=1
    )

    gf_cols = [c for c in g_df.columns if "goals_added_for" in c]
    g_df["goals_added_for"] = g_df.loc[:, gf_cols].sum(axis=1)

    ga_cols = [c for c in g_df.columns if "goals_added_against" in c]
    g_df["goals_added_against"] = g_df.loc[:, ga_cols].sum(axis=1)

    g_df["goals_added_diff"] = g_df["goals_added_for"] - g_df["goals_added_against"]

    return g_df.sort_values("goals_added_diff", ascending=False).reset_index(drop=True)
```

# Most Racing Game of the year


```python
g_df = decorate_goals_added(
    decorate_game_ids(
        decorate_team_ids(
            asa_client.get_player_goals_added(
                leagues="nwsl",
                season_name="2025",
                split_by_games=True,
            )
        )
    )
)

g_df
```

```python
ga_cols = [c for c in g_df.columns if "goals_added_raw" in c]

# g_df
g_df["goals_added_raw_total"] = g_df.loc[:, ga_cols].sum(axis=1)
```

```python
lou_games_df = g_df[g_df.game_desc.str.contains("LOU")]

lou_games_df = (
    lou_games_df.groupby(["team", "game_desc"])
    .goals_added_raw_total.sum()
    .reset_index()
)
```

```python
lou_games_lou = lou_games_df[lou_games_df.team == "LOU"].sort_values(
    "game_desc", ignore_index=True
)


lou_games_notlou = lou_games_df[lou_games_df.team != "LOU"].sort_values(
    "game_desc", ignore_index=True
)

lou_games_notlou
```

```python
lou_games_lou["ga_diff"] = (
    lou_games_lou.goals_added_raw_total - lou_games_notlou.goals_added_raw_total
)
```

```python
lou_games_lou
```

```python
lou_games_xpass = decorate_game_ids(
    decorate_team_ids(
        asa_client.get_team_xpass(
            leagues="nwsl",
            season_name="2025",
            team_ids="eV5DR6YQKn",
            split_by_games=True,
        )
    )
)

lou_games_xpass
```

```python
lou_xgoals = decorate_game_ids(
    asa_client.get_team_xgoals(
        leagues="nwsl",
        season_name="2025",
        team_ids="eV5DR6YQKn",
        split_by_games=True,
        shot_pattern=[
            "Set piece",
            "Corner",
            "Free kick",
            # "Penalty",
            "Fastbreak",
            "Regular",
        ],
    )
)
lou_xgoals
```

```python
most_lou_game = lou_games_xpass[
    [
        "game_id",
        "avg_vertical_distance_difference",
        "avg_vertical_distance_for",
        "game_desc",
    ]
].merge(lou_games_lou[["game_desc", "ga_diff"]], on="game_desc")

most_lou_game = most_lou_game.merge(
    lou_xgoals[["game_desc", "xgoal_difference", "points"]], on="game_desc"
)

most_lou_game["directness_rank"] = most_lou_game.avg_vertical_distance_difference.rank(
    ascending=False, method="min"
)

most_lou_game["ga_rank"] = most_lou_game.ga_diff.rank(ascending=False, method="min")

most_lou_game["npxg_rank"] = most_lou_game.xgoal_difference.rank(
    ascending=False, method="min"
)

most_lou_game["points_rank"] = most_lou_game.points.rank(ascending=False, method="min")

rank_cols = [c for c in most_lou_game.columns if "rank" in c]

most_lou_game["overall_rank"] = most_lou_game[rank_cols].sum(axis=1)

most_lou_game
```

# shots


```python
def get_shots_single_season(leagues="nwsl", season="2024"):
    games_df = asa_client.get_games(leagues=leagues, seasons=season)

    shots_df = asa_client._get_stats(
        game_id=games_df.game_id.to_list(), type="shots", entity="games", leagues="nwsl"
    )

    shots_df["season"] = season

    return shots_df
```

```python
shots_df = decorate_team_ids(get_shots_single_season(season="2025"))
```

```python
shots_df
```

```python
lou_goals = decorate_player_ids(
    decorate_game_ids(shots_df.query("(team=='LOU') & (goal == 1)"))
)

lou_goals
```

```python
shots_df.query("(head == 1)").groupby("team").agg(
    headed_shots=pd.NamedAgg(column="game_id", aggfunc="count"),
    headed_goals=pd.NamedAgg(column="goal", aggfunc="sum"),
    headed_xg=pd.NamedAgg(column="shot_xg", aggfunc="sum"),
    headed_psxg=pd.NamedAgg(column="shot_psxg", aggfunc="sum"),
    headed_shot_players=pd.NamedAgg(column="shooter_player_id", aggfunc="nunique"),
).sort_values("headed_shots", ascending=False).style.format(
    precision=2,
).set_caption(
    "Headed Shots"
)

# Heading
```

```python
shots_df.groupby("team").agg(
    median_xg=pd.NamedAgg(column="shot_xg", aggfunc="median"),
    mean_xg=pd.NamedAgg(column="shot_xg", aggfunc="mean"),
    good_great_shots=pd.NamedAgg(column="shot_xg", aggfunc=lambda x: (x >= 0.15).sum()),
    good_great_rate=pd.NamedAgg(column="shot_xg", aggfunc=lambda x: (x >= 0.15).mean()),
    total_shots=pd.NamedAgg(column="shot_psxg", aggfunc="count"),
    assisted_shots=pd.NamedAgg(
        column="assist_player_id", aggfunc=lambda x: (x != "0").sum()
    ),
).sort_values("assisted_shots", ascending=False)
```

```python
shots_gb = (
    shots_df.query("goal == 1")
    .groupby("team")
    .agg(
        median_xg=pd.NamedAgg(column="shot_xg", aggfunc="median"),
        good_great_shots=pd.NamedAgg(
            column="shot_xg", aggfunc=lambda x: (x >= 0.15).sum()
        ),
        goals=pd.NamedAgg(column="shot_psxg", aggfunc="count"),
        assisted_goals=pd.NamedAgg(
            column="assist_player_id", aggfunc=lambda x: (x != "0").sum()
        ),
    )
)

shots_gb["goal_assist_rate"] = shots_gb.assisted_goals / shots_gb.goals

shots_gb
```
